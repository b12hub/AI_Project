# **Imports and load dataset**

```python
import pandas as pd

# Load the Dataset
df = (pd.read_csv('Dataset/final_dataset.csv')).copy()
```

```python

# 1. Define the target variable (y)
target_column = 'delivery_time_hours'
y = df[target_column].copy()

print(f"Target variable: {target_column}")
print(f"Target shape: {y.shape}")
print()

# 2. Construct the feature matrix (X)
# Remove target column and any columns that could introduce leakage
columns_to_drop = [target_column, 'order_date']  # Target + potential leakage columns

# Create feature matrix by dropping specified columns
X = df.drop(columns=columns_to_drop, errors='ignore')

# 3. Validate separation
print(f"Feature matrix shape: {X.shape}")
print(f"Target vector shape: {y.shape}")
print()
```

## **Validation**

```python
# Confirm column lists
print("Feature columns (first 10):", X.columns.tolist()[:-1])
print(f"Total feature columns: {len(X.columns)}")
print()

# Validation checks
print("VALIDATION CHECKS:")
print(f"1. 'delivery_time_hours' in X: {'delivery_time_hours' in X.columns}")
print(f"2. 'order_date' in X: {'order_date' in X.columns}")

# Additional check for any ID-like columns
id_like_patterns = ['id', 'ID', 'Id', 'order_id', 'customer_id', 'tracking_id']
id_like_in_columns = any(pattern in col.lower() for col in X.columns for pattern in id_like_patterns)
print(f"3. Any ID-like columns in X: {id_like_in_columns}")
```

## **Final Asseration**

```python
assert 'delivery_time_hours' not in X.columns, "Target leakage detected: delivery_time_hours in X"
assert 'order_date' not in X.columns, "Temporal leakage detected: order_date in X"
print("\n✅ All validation checks passed!")

# Show column information
print("\n" + "="*60)
print("FEATURE COLUMNS BY TYPE:")
print("="*60)

# Categorical columns (will need encoding later)
categorical_cols = X.select_dtypes(include=['object']).columns.tolist()
print(f"Categorical columns ({len(categorical_cols)}): {categorical_cols}")

# Numerical columns
numerical_cols = X.select_dtypes(include=['int64', 'float64']).columns.tolist()
print(f"\nNumerical columns ({len(numerical_cols)}): {numerical_cols}")

# Boolean columns
bool_cols = X.select_dtypes(include=['bool']).columns.tolist()
print(f"\nBoolean columns ({len(bool_cols)}): {bool_cols}")


# df.to_csv('Dataset/feature_target.csv', index=False)

print("\n" + "="*60)
print("✅ BLOCK 2.4.1 COMPLETE: Feature/target separation is safe for splitting.")
print(f"   - X shape: {X.shape}")
print(f"   - y shape: {y.shape}")
print("="*60)
```
