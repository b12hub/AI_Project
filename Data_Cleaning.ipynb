## **Load the raw dataset**

```python
import pandas as pd
import numpy as np


print("Loading raw dataset...")
df = pd.read_csv('dataset/raw_data.csv')

print(f"Input: {df.shape[0]} rows × {df.shape[1]} columns")
```

## **STEP 1: Drop non-predictive identifier**


```python
df_clean = df.drop('order_id', axis=1)
print("✅ Dropped order_id (non-predictive identifier)")
```

## **STEP 2: Type enforcement**
### **Convert order_date to datetime**


```python
df_clean['order_date'] = pd.to_datetime(df_clean['order_date'])

# Enforce numeric types
dtype_mapping = {
    'distance_km': 'float64',
    'order_hour': 'int64',
    'temperature': 'float64',
    'delivery_time_hours': 'float64'
}

for col, dtype in dtype_mapping.items():
    df_clean[col] = df_clean[col].astype(dtype)

print("✅ Enforced correct data types")
```

## **STEP 3: Hard validation rules**


```python
validation_rules = {
    'order_hour_range': ((df_clean['order_hour'] >= 0) & (df_clean['order_hour'] <= 23)).all(),
    'distance_positive': (df_clean['distance_km'] > 0).all(),
    'delivery_positive': (df_clean['delivery_time_hours'] > 0).all(),
    'different_cities': (df_clean['origin_city'] != df_clean['destination_city']).all()
}

for rule_name, passed in validation_rules.items():
    if not passed:
        raise ValueError(f"Validation rule failed: {rule_name}")

print("✅ All hard validation rules passed")
```

## **STEP 4: Categorical normalization**


```python
categorical_cols = df_clean.select_dtypes(include=['object']).columns.tolist()

for col in categorical_cols:
    df_clean[col] = df_clean[col].str.strip().str.lower()

print("✅ Normalized categorical variables")
```

## **STEP 5: Remove duplicates**

```python
initial_count = len(df_clean)
df_clean = df_clean.drop_duplicates()
removed_count = initial_count - len(df_clean)

print(f"✅ Removed {removed_count} duplicate rows")
```

## **STEP 6: Verify missing values**

```python
if df_clean.isnull().sum().sum() > 0:
    raise ValueError("Missing values found in cleaned dataset")

print("✅ No missing values detected")

# Save cleaned dataset
output_path = 'dataset/clean_data_v1.csv'
df_clean.to_csv(output_path, index=False)

print(f"✅ Saved cleaned dataset to {output_path}")
print(f"Output: {df_clean.shape[0]} rows × {df_clean.shape[1]} columns")

# Final verification
print("\nData cleaning completed successfully!")
print("Dataset is ready for feature engineering (BLOCK 2.3)")
```
