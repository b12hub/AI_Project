# **STEP 1: Load & sanity check**

```python
import pandas as pd

# Load feature-engineered Dataset
df = pd.read_csv("Dataset/feature_data_v1.csv")

# 1. Re-check missing values
assert df.isnull().sum().sum() == 0, "Missing values detected"

# 2. Normalize categorical columns
cat_cols = [
    'origin_city', 'destination_city',
    'vehicle_type', 'weekday',
    'weather', 'traffic_level'
]

for col in cat_cols:
    df[col] = df[col].str.strip().str.lower()

# 3. Enforce correct data types
df['order_date'] = pd.to_datetime(df['order_date'])
df['order_hour'] = df['order_hour'].astype(int)
df['distance_km'] = df['distance_km'].astype(float)
df['temperature'] = df['temperature'].astype(float)
df['delivery_time_hours'] = df['delivery_time_hours'].astype(float)

# Binary features
binary_cols = ['is_weekend', 'is_peak_hour', 'is_night']
for col in binary_cols:
    df[col] = df[col].astype(bool)

# 4. Remove full duplicates (defensive)
df = df.drop_duplicates()

# 5. Final sanity check
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Dataset clean and stable.")

# Save
df.to_csv("Dataset/clean_data_v2.csv", index=False)
print("BLOCK 2.3.2 COMPLETED")
```
