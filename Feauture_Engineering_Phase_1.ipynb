## **LOAD DATA**


```python
import pandas as pd
df = pd.read_csv('dataset/clean_data_v1.csv')

original_row_count = len(df)
```

## **DATE CONVERSION**


```python
df['order_date'] = pd.to_datetime(df['order_date'])
```

## **TEMPORAL FEATURES**

```python
# Day of month (1–31)
df['day_of_month'] = df['order_date'].dt.day

# ISO week of year (cast to int for ML compatibility)
df['week_of_year'] = df['order_date'].dt.isocalendar().week.astype(int)

# Month (1–12)
df['month'] = df['order_date'].dt.month

# Weekend flag
df['is_weekend'] = df['weekday'].isin(['saturday', 'sunday']).astype(int)

# Peak traffic hours: 8–10, 17–19
peak_hours = list(range(8, 11)) + list(range(17, 20))
df['is_peak_hour'] = df['order_hour'].isin(peak_hours).astype(int)

# Night hours: 22–05
night_hours = list(range(22, 24)) + list(range(0, 6))
df['is_night'] = df['order_hour'].isin(night_hours).astype(int)
```

## **VALIDATION CHECKS**

```python
assert len(df) == original_row_count, "❌ Row count changed!"
```

## **SAVE OUTPUT (MANDATORY)**

```python
df.to_csv('dataset/feature_data_v1.csv', index=False)
```
